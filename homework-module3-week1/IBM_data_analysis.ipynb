```python
revenue_mean = data['Revenue (Millions)'].mean()
data['Revenue (Millions)'].fillna(revenue_mean, inplace = True)
```

```python
def group_rating(rating):
    if rating >= 7.5:
        return 'Good'
    elif rating >= 6.0:
        return 'Average'
    else:
        return 'Bad'

data['Rating_category'] = data['Rating'].apply(group_rating)
data [['Title','Director','Rating','Rating_category']].head(5)
```
